python data type: ARRAY(VETTORI), MATRIX 

```python
a = np.array([4,6,7,8,3,4,5,6,7,8,9,5,3,2,1,0])
estratto27 = a[2:8]  # Estrae gli elementi dall'indice 2 all'indice 7
estratto5fine = a[5:-1]  # Estrae gli elementi dall'indice 5 al penultimo elemento
print(f"Elementi estratti da 2 a 7: {estratto27}")
print(f"Elementi estratti da 5 a penultimo elemento: {estratto5fine}")
print(f"Elementi estratti da 5 a fine: {a[5:]}")  # da 5 a fine dell'array
print(f"Elementi estratti da 5 a fine: {a[:3]}")  # da inizio a 3
```

```python
import numpy as np

# Ricreiamo la nostra matrice 2D con le due righe precedenti
riga_1 = np.array([10, 20, 30, 40, 50, 60, 70, 80])
riga_2 = riga_1 * 2
riga_3 = np.array([10, 11, 12, 13])
matrice = np.array([riga_1,riga_2,riga_2])
matrice2 = np.array([[1,2],[3,4]])
matrice3 = np.array([riga_3]).reshape(2,2)


print("Matrice:\n", matrice)
print("\nMatrice2:\n", matrice2)
print("\nMatrice3:\n", matrice3)


print("\nMatrice2[2,1]:\n", matrice2[2-1, 1-0]) # gli indici partono da 0, quindi per la seconda riga e prima colonna usiamo [1,0]

# 1. Estrarre un singolo elemento specifico: matrice[riga, colonna]
# Vogliamo il numero nella 1ª riga (indice 0) e 3ª colonna (indice 2) -> il numero 30
print("Elemento in riga 0, colonna 2:", matrice[0, 2])

# 2. Estrarre un'intera riga (es. tutta la seconda riga)
# Il simbolo ':' significa "prendi tutto" per quella dimensione
print("Intera seconda riga:", matrice[1, :])

# 3. Estrarre un'intera colonna (es. la terza colonna di entrambe le righe)
print("Terza colonna di tutte le righe:", matrice[:, 2])
```

python data type: TUPLE

```python
import numpy as np

# le tuple sono immutabili, non possono essere modificate dopo la creazione
# sono come le liste, ma non possono essere cambiate
# sono utili per rappresentare dati che non devono essere alterati, come coordinate o configurazioni fisse
# sono come le CONST in C/C++ o final in Java

miaTupla = (1, 2, 3)
miaLista = [4, 5, 6]


matrice = np.array([[1, 2, 3], [4, 5, 6]])
matrice2 = np.array([miaTupla, miaLista])  # Creiamo una matrice 2D con una tupla e una lista

print("Matrice creata elenco\n", matrice)
print("Matrice creata da tupla e lista:\n", matrice2)

```

linear algebra: multiplication, eigenvalues(autovalori), eigenvector (autovettori)

```python
import numpy as np

# Creiamo una matrice di test 2x2
A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[2, 0], 
              [1, 2]])

# 1. Prodotto matriciale con l'operatore @
prodotto = A @ B
print("Prodotto matriciale (A @ B):\n", prodotto)

# 2. Matrice trasposta con .T
trasposta_A = A.T
print("\nMatrice A trasposta (A.T):\n", trasposta_A)

# 3. Matrice inversa con np.linalg.inv()
inversa_A = np.linalg.inv(A)
print("\nMatrice A inversa (np.linalg.inv):\n", inversa_A)

# 4. Autovalori e autovettori con np.linalg.eig()
# La funzione restituisce due elementi: un vettore con gli autovalori e una matrice con i relativi autovettori
autovalori, autovettori = np.linalg.eig(A)

autovettore_1 = autovettori[:, 0]
autovettore_2 = autovettori[:, 1]


print("\nAutovalori di A:\n", autovalori)
print("\nAutovettori di A:\n", autovettori)
print(f"\nPrimo autovettore di A: {autovettore_1} e autovalore associato: {autovalori[0]}")
print(f"Secondo autovettore di A: {autovettore_2} e autovalore associato: {autovalori[1]}\n")

```
