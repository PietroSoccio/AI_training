linear algebra: multiplication, eigenvalues(autovalori), eigenvector (autovettori)

```python
import numpy as np

# Creiamo una matrice di test 2x2
A = np.array([[1, 2], 
              [3, 4]])

B = np.array([[2, 0], 
              [1, 2]])

# 1. Prodotto matriciale con l'operatore @
prodotto = A @ B
print("Prodotto matriciale (A @ B):\n", prodotto)

# 2. Matrice trasposta con .T
trasposta_A = A.T
print("\nMatrice A trasposta (A.T):\n", trasposta_A)

# 3. Matrice inversa con np.linalg.inv()
inversa_A = np.linalg.inv(A)
print("\nMatrice A inversa (np.linalg.inv):\n", inversa_A)

# 4. Autovalori e autovettori con np.linalg.eig()
# La funzione restituisce due elementi: un vettore con gli autovalori e una matrice con i relativi autovettori
autovalori, autovettori = np.linalg.eig(A)

autovettore_1 = autovettori[:, 0]
autovettore_2 = autovettori[:, 1]


print("\nAutovalori di A:\n", autovalori)
print("\nAutovettori di A:\n", autovettori)
print(f"\nPrimo autovettore di A: {autovettore_1} e autovalore associato: {autovalori[0]}")
print(f"Secondo autovettore di A: {autovettore_2} e autovalore associato: {autovalori[1]}\n")

```

Programmazione a oggetti OOP object-oriented

```python
import numpy as np

class DataManager:
    """Classe per la gestione e il salvataggio di dati e file di log."""
    
    def __init__(self, nome_progetto="AI_Training"):
        self.nome_progetto = nome_progetto
        print(f"Inizializzato DataManager per il progetto: {self.nome_progetto}")

    def salva_numpy(self, dati, filename="dati_ml.npy"):
        """Salva una matrice/array in formato binario NumPy."""
        np.save(filename, dati)
        print(f"[OK] Dati salvati con successo in binario: {filename}")

    def carica_numpy(self, filename="dati_ml.npy"):
        """Carica dati da un file binario NumPy."""
        dati = np.load(filename)
        print(f"[OK] Dati caricati da: {filename}")
        return dati

    def salva_csv(self, dati, filename="dati_ml.csv", header="C1,C2,C3"):
        """Salva i dati in formato CSV tabulare."""
        np.savetxt(filename, dati, delimiter=',', header=header, comments='')
        print(f"[OK] Dati esportati in CSV: {filename}")

    def scrivi_log(self, messaggio, filename="registro_training.txt"):
        """Aggiunge una riga di testo a un file di registro."""
        with open(filename, 'a') as f:
            f.write(messaggio + "\n")
        print(f"[OK] Log registrato su: {filename}")

    def leggi_log(self, filename="registro_training.txt"):
        """Legge e stampa il contenuto del registro."""
        try:
            with open(filename, 'r') as f:
                contenuto = f.read()
                print(f"\n--- Contenuto di {filename} ---")
                print(contenuto)
                print("---------------------------------")
        except FileNotFoundError:
            print(f"Il file {filename} non esiste ancora.")


# --- ESEMPIO DI UTILIZZO ---

# 1. Istanziamo l'oggetto
manager = DataManager("Modulo_Base_Python")

# 2. Creiamo una matrice di esempio
matrice_dati = np.array([[1.5, 2.0, 3.1], [4.2, 5.5, 6.0]])

# 3. Usiamo i metodi dell'oggetto
manager.salva_numpy(matrice_dati, "matrice_peso.npy")
manager.salva_csv(matrice_dati, "matrice_peso.csv")

# 4. Scriviamo un log delle attività
manager.scrivi_log("Completato test di salvataggio oggetti a blocchi.")

# 5. Leggiamo il registro
manager.leggi_log()
```

Gestione delle eccezioni try-except

```python
import numpy as np

class DataManager:
    """Classe per la gestione dei dati con gestione avanzata degli errori."""
    
    def __init__(self, project_name="AI_Training"):
        self.project_name = project_name

    def carica_numpy(self, filename="dati_ml.npy"):
        """Carica dati da un file binario gestendo l'errore se il file non esiste."""
        try:
            # Tentativo di eseguire l'operazione "rischiosa"
            dati = np.load(filename)
            print(f"[OK] Dati caricati correttamente da: {filename}")
            return dati
            
        except FileNotFoundError:
            # Cosa fare se il file non viene trovato
            print(f"[ERRORE] Il file '{filename}' non esiste. Controlla il percorso o salva prima i dati.")
            return None
            
        except Exception as e:
            # Cattura qualsiasi altro tipo di errore imprevisto
            print(f"[ERRORE IMPREVISTO] Si è verificato un problema: {e}")
            return None


# --- ESEMPIO DI UTILIZZO ---
manager = DataManager("Test_Eccezioni")

# Proviamo a caricare un file che NON esiste
print("Tentativo di caricamento di un file inesistente:")
dati_inesistenti = manager.carica_numpy("file_fantasma.npy")
```
